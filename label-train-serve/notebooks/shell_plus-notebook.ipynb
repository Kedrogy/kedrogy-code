````python
import os

from label_train_serve.models import DjangoDataset, DjangoModel, DjangoLastDataset


os.environ["DJANGO_ALLOW_ASYNC_UNSAFE"] = "True"
````

````python
that_dataset = DjangoDataset.objects.filter(dataset_name="news_headlines").first()

that_dataset
````

````python
def update_latest_dataset(dataset_id):
    found = list(DjangoLastDataset.objects.all())
    if len(found) > 0:
        DjangoLastDataset.objects.all().delete()
    latest = DjangoLastDataset(dataset_id=dataset_id).save()
    return latest
````

````python
def get_latest_dataset():
    return list(DjangoLastDataset.objects.all())[0]
````

````python
update_latest_dataset(that_dataset.id)
````
