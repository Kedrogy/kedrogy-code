```python
import json

import requests


# curl -H "Content-Type: application/json" http://0.0.0.0:8888/predict -d '{"text":"Hello, world"}'

url = "http://0.0.0.0:8888/predict"

resp = requests.post(
    url=url,
    data=json.dumps(
        {
            "text": "Hello, world",
        }
    ),
)
predicted_class_id = resp.json()["predicted_class_id"]

predicted_class_id
```
